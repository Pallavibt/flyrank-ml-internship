## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 pages across 32 pseudonymized clients. It marks 16,262 pages (54.2%) as declining. My strict initial rule finds only 17 visible, stale review candidates (0.1% of pages), and 94.1% of those are marked declining. This is encouraging but not proof that the rule is best: the candidate set is very small, and the decline flag is a derived label. Over the next weeks, I will test whether a transparent ranking using several observable signals produces a more useful review queue.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Pallavibt/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

declining = df["trend_direction"].eq("down")
review_candidate = (
    (df["days_since_last_update"] >= 180)
    & (df["impressions_90d"] >= 500)
)

print(f"Pages: {len(df):,}")
print(f"Clients: {df['client_id'].nunique():,}")
print(f"Pages marked declining: {declining.sum():,} ({declining.mean():.1%})")
print(
    f"Visible, stale review candidates: {review_candidate.sum():,} "
    f"({review_candidate.mean():.1%})"
)
print(
    "Declining rate among review candidates: "
    f"{declining[review_candidate].mean():.1%}"
)
```
